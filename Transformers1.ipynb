```python
from transformers import pipeline

# Sentiment analysis — pretrained on millions of reviews
sentiment = pipeline("sentiment-analysis")

# Test it:
results = sentiment([
    "I absolutely loved working in Perth amazing city",
    "Terrible experience never coming back horrible",
    "Not bad actually better than I expected",  # tricky one!
    "The food was okay nothing special average"
])

for result in results:
    print(result)
```

```python
from transformers import pipeline

# 3. Text Generation (more useful than summarisation)
generator = pipeline("text-generation",
                     model="distilgpt2")

result = generator(
    "The future of AI engineering in Perth is",
    max_length=50,
    num_return_sequences=1
)
print("Generated:", result[0]['generated_text'])

# 4. Question Answering
qa = pipeline("question-answering")

context = """
Perth is located in Western Australia and is the fourth largest city
in Australia. The city is known for its technology sector, mining industry,
and beautiful beaches. The average salary for an AI engineer in Perth
is around $120,000 to $150,000 per year.
"""

questions = [
    "Where is Perth located?",
    "What is the average salary for an AI engineer in Perth?",
    "What is Perth known for?"
]

for question in questions:
    result = qa(question=question, context=context)
    print(f"\nQ: {question}")
    print(f"A: {result['answer']} (confidence: {result['score']:.2%})")
```
