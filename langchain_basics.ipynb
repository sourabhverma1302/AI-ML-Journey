```python
from transformers import pipeline
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import PromptTemplate

# Same pipeline that works
pipe = pipeline(
    "text-generation",
    model="distilgpt2",
    max_new_tokens=50
)

llm = HuggingFacePipeline(pipeline=pipe)

# Add PromptTemplate
template = PromptTemplate(
    input_variables=["topic"],
    template="Machine learning concept: {topic} means"
)

# Chain them
chain = template | llm

# Test with 3 topics
topics = ["overfitting", "neural networks", "random forest"]

for topic in topics:
    result = chain.invoke({"topic": topic})
    print(f"Topic: {topic}")
    print(f"Output: {result}")
    print()
```

```python
pipe = pipeline(
    "text-generation",
    model="distilgpt2",
    max_new_tokens=50
)

template=PromptTemplate(
    input_variables=["topic","target_audience"],
    template="teach about {topic} according to the {target_audience}"
)

llm=HuggingFacePipeline(pipeline=pipe)

prompts=[
    {"topic":"AI","target_audience":"10 year old"},
    {"topic":"Periods","target_audience":"15 year old boy"},
    {"topic":"Rainbow","target_audience":"fish"}
]

chain=template | llm

for prompt in prompts:
  output=chain.invoke(prompt)
  print(output)
```

```python
pipe = pipeline(
    "text-generation",
    model="distilgpt2",
    max_new_tokens=50
)

topic_template=PromptTemplate(
    input_variables=["topic"],
    template="explain about this {topic}"
)

explain_template=PromptTemplate(
    input_variables=["explanation"],
    template="give us one question according to this {explanation}"
)

llm=HuggingFacePipeline(pipeline=pipe)

chain1=topic_template | llm
chain2=explain_template | llm

explanation=chain1.invoke({"topic":"car"})
print(explanation)
output=chain2.invoke({"explanation":explanation})
print(output)
```
